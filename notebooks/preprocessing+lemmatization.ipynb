```python
import os
import re
import nltk
from nltk.corpus import stopwords
!pip install pymorphy2==0.8
from pymorphy2 import MorphAnalyzer
```

```python
# Скачиваем стоп-слова для русского языка
nltk.download('stopwords')
stop_words = set(stopwords.words('russian'))

# Инициализация анализатора
morph = MorphAnalyzer()

def clean_text(text, remove_stopwords=True):
#Функция для предварительной очистки текста (предобработка)
    text = re.sub(r'[^а-яА-ЯёЁ\s]', ' ', text)  # Удаляем всё, кроме букв и пробелов
    text = re.sub(r'\s+', ' ', text)  # Удаляем лишние пробелы
    text = text.lower().strip()  # Приводим к нижнему регистру и обрезаем пробелы

    if remove_stopwords:
      words = text.split()
      words = [word for word in words if word not in stop_words]
      text = ' '.join(words)
    return text

def lemmatize_text(text, remove_stopwords=True):
#Функция для лемматизации текста
    words = text.split()
    lemmatized = []
    for word in words:
        if remove_stopwords and word in stop_words:
            continue
        lemma = morph.parse(word)[0].normal_form
        lemmatized.append(lemma)
    return ' '.join(lemmatized)

def process_file(file_path, output_folder):
#Обработка одного файла
    with open(file_path, 'r', encoding='utf-8') as f:
        text = f.read()

    # 1. Предобработка (очистка без лемматизации)
    text_clean = clean_text(text)
    clean_filename = f"clean_{os.path.basename(file_path)}"
    clean_path = os.path.join(output_folder, clean_filename)

    with open(clean_path, 'w', encoding='utf-8') as f:
        f.write(text_clean)

    # 2. Лемматизация (очистка + приведение к леммам)
    text_lemma = lemmatize_text(text_clean)
    lemma_filename = f"lemma_{os.path.basename(file_path)}"
    lemma_path = os.path.join(output_folder, lemma_filename)

    with open(lemma_path, 'w', encoding='utf-8') as f:
        f.write(text_lemma)

    print(f"Обработан файл: {os.path.basename(file_path)}")

def main():
    input_folder = '/content/drive/My Drive/SFU 4/Ex ml/1_подготовленный_корпус/corpus/'  # Путь к исходным файлам
    output_folder = '/content/drive/My Drive/SFU 3/Ex ml/2_морфологический_анализ_с_использованием_UDPipe/Lemma/'  # Путь для результатов

    # Создаем папку для результатов, если её нет
    os.makedirs(output_folder, exist_ok=True)

    # Обрабатываем все txt-файлы в папке
    for filename in os.listdir(input_folder):
        if filename.endswith('.txt'):
            file_path = os.path.join(input_folder, filename)
            process_file(file_path, output_folder)

if __name__ == '__main__':
    main()
```
